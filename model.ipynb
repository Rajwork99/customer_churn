This customer churn dataset analyzes why customers discontinue telecommunications services. It contains customer demographics, account details, subscribed services, contract types, billing information, and tenure. The target variable identifies whether a customer has churned, enabling predictive modeling and insights to support retention strategies and improve customer satisfaction..


## Problem Statement

You are the data scientist at a telecom company named “Neo” whose customers are churning out to its competitors. You have to analyze the data of your company and find insights and stop your customers from churning out to other
telecom companies.


## Dataset Information


1. **customerID** – Unique identifier for each customer.
2. **gender** – Customer gender: Male or Female.
3. **SeniorCitizen** – Indicates whether the customer is a senior citizen: 1 or 0.
4. **Partner** – Indicates whether the customer has a partner.
5. **Dependents** – Indicates whether the customer has dependents.
6. **tenure** – Number of months the customer has stayed with the company.
7. **PhoneService** – Indicates whether the customer uses phone service.
8. **MultipleLines** – Indicates whether the customer has multiple phone lines.
9. **InternetService** – Type of internet service: DSL, Fiber optic, or No.
10. **OnlineSecurity** – Indicates whether the customer has online security service.
11. **OnlineBackup** – Indicates whether the customer has online backup service.
12. **DeviceProtection** – Indicates whether the customer has device protection service.
13. **TechSupport** – Indicates whether the customer has technical support service.
14. **StreamingTV** – Indicates whether the customer subscribes to TV streaming.
15. **StreamingMovies** – Indicates whether the customer subscribes to movie streaming.
16. **Contract** – Contract duration: month-to-month, one year, or two years.
17. **PaperlessBilling** – Indicates whether the customer uses paperless billing.
18. **PaymentMethod** – Customer’s payment method.
19. **MonthlyCharges** – Amount charged to the customer monthly.
20. **TotalCharges** – Total amount charged to the customer.
21. **Churn** – Target variable indicating whether the customer left the company.
